```python
# Parent class

# Instance Attributes: Attributes that are specific to each instance of the class. Defined in the __init__ method.
# Class Attributes: Attributes that are shared among all instances of the class. Defined outside any instance methods.

class Book:
    # Class attribute
    total_books = 0

    # Constructor
    # The __init__ method is a special method in Python that initializes a new instance of the class.
    # It is called automatically when a new instance is created.

    def __init__(self, title, author, publication_year):
        # Instance attributes
        self.title = title
        self.author = author
        self.publication_year = publication_year
        self.__is_available = True  # Private attribute to represent book availability

        # Update class attribute
        Book.total_books += 1

    # Instance method to display book details as this works on instances and take self as first parameter,
    # its value differs for every instance
    def display_info(self):
        """Displays book details."""
        status = "Available" if self.__is_available else "Checked Out"
        print(f"Title: {self.title}, Author: {self.author}, Year: {self.publication_year}, Status: {status}")

    # OVERLOADED METHOD
    # By setting 'user' to None, we allow the method to be called in two ways.
    def checkout(self, user=None):
        """
        Overloaded-style method:
        1. book.checkout() -> Standard checkout
        2. book.checkout("Alice") -> Checkout to a specific user
        """
        if self.__is_available:
            self.__is_available = False
            if user:
                print(f"The book '{self.title}' has been checked out to {user}.")
            else:
                print(f"The book '{self.title}' has been checked out.")
        else:
            print(f"The book '{self.title}' is already checked out.")

    # Instance method to return a book
    def return_book(self):
        """Returns the book to the library."""
        self.__is_available = True
        print(f"The book '{self.title}' has been returned to the library.")

# Instance Methods:
# Functions that belong to the class and usually operate on instance attributes(different for every instance, defined inside instance methods). The first parameter is self, which represents the instance.

# Class Methods:
# Functions that operate on class attributes(same for entire class, defined inside class). The first parameter is cls, which represents the class itself. Defined using the @classmethod decorator.

# Static Methods:
# Functions that neither operate on instance attributes nor class attributes. Defined using the @staticmethod decorator.

    # Class method to get the total number of books, this works on class attributes and has to use a decorator @classmethod
    @classmethod
    def get_total_books(cls):
        return cls.total_books

    # Static method to validate a publication year, this is a general method which doesn't depend on intance or class variables
    @staticmethod
    def is_valid_year(year):
        """Static method to check if the publication year is valid."""
        return year > 0

    # make a method overloading example here
    def
```

```python
# Child class inheriting from Book
# A mechanism that allows a class (child class) to inherit attributes and methods from another class (parent class).

class EBook(Book):
    def __init__(self, title, author, publication_year, file_size):
        # Call the parent class constructor
        super().__init__(title, author, publication_year)
        # Additional instance attribute for EBook
        self.file_size = file_size

    # Overriding the display_info method to include file size
    def display_info(self):
        super().display_info()  # Call the parent class method
        print(f"File Size: {self.file_size}MB")

# Encapsulation: Restricting access to some of an object's components.
# This is often achieved by prefixing an attribute or method with a single or double underscore (_ or __).
```

```python
# Example Usage
# Creating instances of Book and EBook
book1 = Book("1984", "George Orwell", 1949)
book2 = EBook("Digital Fortress", "Dan Brown", 1998, 5)
```

```python
# Displaying book information
book2.display_info()
book1.display_info()
```

```python
# Checking out a book
book1.checkout()
book1.checkout()  # Attempt to check out again
```

```python
# Returning a book
book1.return_book()
```

```python
# Using class method
print(f"Total books in library: {Book.get_total_books()}")

# Using static method
print(f"Is 2022 a valid year? {Book.is_valid_year(2022)}")
```
