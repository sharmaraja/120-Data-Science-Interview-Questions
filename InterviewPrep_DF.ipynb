**Loc and iloc**

```python
# loc is label of index based indexing df.loc[1] will give row where index=1
# iloc is integer based indexing irrespective of their labels df.iloc[1] will give 1st row index whatever be the value
# Used for looping over the dataframe-computationaly very expensive
```

```python
print('loc\n',df['col3'].loc[2],'\niloc\n',df['col3'].iloc[4])
# as index value and index number are same we get same results
```

```python
# Sample DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve','check'],
        'Age': [25, 30, 35, 40, 45,59]}
df = pd.DataFrame(data)

# Define bin edges and labels
bins = [0, 30, 40, 50]  # Bin edges
labels = ['Young', 'Middle-aged', 'Old']  # Bin labels

# Apply `cut` to create a new column 'Age Group'
df['Age Group'] = pd.cut(df['Age'], bins=bins, labels=labels)

df
```

```python
df['AG'] = df['Age'].apply(lambda x: 'Young' if x<=30 else 'Mid' if 30<x<=40 else 'Old')
df
```

```python
import pandas as pd

# Sample dataset
data = {
    'Region': ['East', 'East', 'West', 'West', 'North', 'North', 'South', 'South'],
    'Product': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Sales': [234, 150, 210, 120, 345, 200, 500, 420],
    'Quantity': [10, 15, 8, 12, 11, 16, 9, 14]
}

df = pd.DataFrame(data)
df.head()
```

```python
df_summ = df.groupby(by=['Region','Product']).agg({'Sales':'sum','Quantity':'mean'})
df_summ.head(15)
```

```python
# Merge and Join
df_1 = pd.DataFrame({'A': ['A0', 'A1', 'A2', 'A3'],
                     'B': ['B0', 'B1', 'B2', 'B3'],
                     'C': ['C', 'D', 'E', 'F']},
                    index=[0, 1, 2, 3])

df_2 = pd.DataFrame({'A': ['A4', 'A5', 'A6', 'A7'],
                     'B': ['B4', 'B5', 'B6', 'B7'],
                     'C': ['G', 'H', 'I', 'J']},
                    index=[4, 5, 6, 7])
```

```python
df_3 = pd.merge(df_1, df_2, how='outer', on='A')
df_3.head()
```

```python
# Pivot table with multiple indices
pivot_table_multi_index = pd.pivot_table(df,
                                         values=['Sales', 'Quantity'],
                                         index=['Region', 'Product'],
                                         aggfunc={'Sales': 'sum', 'Quantity': 'mean'},
                                         fill_value=0)

print(pivot_table_multi_index)
```

```python
d = {'id':[1,2,3],
     'name':['Cooler', 'Facewash','Lays FOOD'],
     'Sector':['Electronics','Consumer Durable','Eatable']}
df = pd.DataFrame(d)
df.head()
```

```python
df['code'] = df['Sector'].apply(lambda x: 'ELEC' if x=='Electronics' else 'CD' if x=='Consumer Durable' else 'FOOD' if x=='Eatable' else 'NA')
df.head()
```

```python
df['code_map'] = df['Sector'].map(lambda x: 'ELEC' if x=='Electronics' else 'CD' if x=='Consumer Durable' else 'FOOD' if x=='Eatable' else 'NA')
df.head()
```

```python
df['code_where'] = ''
df['code_where'] = np.where(df['Sector']=='Electronics', 'ELEC', df['code_where'])
df['code_where'] = np.where(df['Sector']=='Consumer Durable', 'CD', df['code_where'])
df['code_where'] = np.where(df['Sector']=='Eatable', 'Food', df['code_where'])

df.head()
```

```python
df['code_repl'] = df['Sector'].replace('Electronics','ELEC')
df['code_repl'] = df['code_repl'].replace('Consumer Durable','CD')
df['code_repl'] = df['code_repl'].replace('Eatable','FOOD')
df.head()
```
