```python
# Flatten nested dictionary and convert to DataFrame
def flatten_dict(d, parent_key='', sep='_'):
    items = []
    for k, v in d.items():
        new_key = f"{parent_key}{sep}{k}" if parent_key else k
        if isinstance(v, dict):
            items.extend(flatten_dict(v, new_key, sep).items())
        elif isinstance(v, list):
            items.append((new_key, ', '.join(v)))
        else:
            items.append((new_key, v))
    return dict(items)
```

```python
# Process all records
flattened_data = [flatten_dict(record) for record in nested]
flattened_data
```

### Error Handling

```python
def error_hand(a,b):
    '''Return a/b'''
    try:
        # Code that may raise an exception
        x = a/b  # This will raise a "ZeroDivisionError"
        print(a/b)
    except ZeroDivisionError as e:
        # Handling the specific exception
        print("An error occurred:", e)
    except Exception as e:
        # Handling a more general exception (if needed)
        print("A different error occurred:", e)
    else:
        # Code to execute if no exception was raised
        print("No exceptions occurred.")
    finally:
        # Code to execute whether there was an exception or not
        print("This code block always runs.")
```

**Sorting and Searching**

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    print('Using bubble sort\n',arr)
```

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    print('Using selection sort\n',arr)
```

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    print('Insertion sort\n',arr)
```

```python
x = 10
y = x
print(id(x),'\n',id(y))
```

```python
x = 20
print(id(x),'\n',id(y))

# string, tuple, or integer--> immutable--> changes in these makes a new object hence new id
# list, dictionary, or set --> mutable --> changes in these does not make a new object but only updates the old one hence same id
```

Key Points:



*   Variables store references to objects in memory.
*   Assigning a variable to another variable creates another reference to the same object.
*   Reassigning a variable can create a new object and update the variable's reference.
*   Modifying mutable objects (like lists) through a reference modifies the original object.
