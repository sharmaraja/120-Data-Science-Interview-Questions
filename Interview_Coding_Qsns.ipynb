```python
# Count the occurence and index position for the srting
s = 'dhkjahdshdkjshdkdhkjsahdkbxb'

# Method 1: Using regular dictionary
count_dict = {}
index_dict = {}

# gives index and value in that order
for i, char in enumerate(s):
    if char not in count_dict:
        count_dict[char] = 1
        index_dict[char] = [i]
    else:
        count_dict[char] += 1
        index_dict[char].append(i)

print("Character counts:")
print(count_dict)
print("\nCharacter positions:")
print(index_dict)
```

```python
# 1) Flattening a list without built-in function (
l = [1,[2,[3,[4,5],6],7],8]
l1 = []

def flat(ip_l,op_l):
  for i in ip_l:
    if type(i)==list:
      # In Python, type(i) returns a type object, not a string.
      flat(i,op_l)
    else:
      op_l.append(i)
  return op_l

print(flat(l,l1))
```

```python
# 7) Create fucntion to remove char that is present more than once.
x = ["aaabccdefgh"]

def remove_repeating_chars(s):
    # 1. Count frequencies of all characters
    counts = {}
    for char in s:
        counts[char] = counts.get(char, 0) + 1

    # 2. Build a new string using only chars that appeared exactly once
    result = ""
    for char in s:
        if counts[char] == 1:
            result += char

    print(counts)
    return result

for i in x:
  print(remove_repeating_chars(i))
```

```python
# 7) Define a Python function that takes two different strings as arguments and evaluates whether the strings have the same characters
# and frequencies. If both conditions are met, return True; otherwise, return False.

def check(a,b):
  dict_a={}
  dict_b={}
  if len(a) != len(b):
    return('False')
  if set(a) != set(b) :
    return('False')
  else:
    for i in a:
      dict_a[i] = dict_a.get(i,0) + 1
    for i in b:
      dict_b[i] = dict_b.get(i,0) + 1
    if dict_a == dict_b:
      return('True')

print(check('abcde','edcbaa'))
```

```python
def are_anagrams(s1, s2):
    return sorted(s1) == sorted(s2)

# Testing the function
print(are_anagrams("listen", "silent"))  # Output: True
print(are_anagrams("hello", "world"))  # Output: False
print(are_anagrams('abcde','edcbaa'))  # Output: True
print(are_anagrams('abcde','aedcb'))  # Output: False
```

```python
def first_non_repeating(s):
    count = {}
    for char in s:
        count[char] = count.get(char, 0) + 1
    for char in s:
        if count[char] == 1:
            return char
    return None
```

```python
print(first_non_repeating("swiss"))  # Output: 'w'
```

```python
def lng_ss(s):
  temp=''
  res=''
  length=0
  max=0
  for i in s:
    if i not in  temp:
      temp+=i
      length+=1
    else:
      if length>max:
        max=length
        res = temp
      temp=''
      length=0
  return res
```

```python
print(lng_ss('abbefghhabcbb'))
```
